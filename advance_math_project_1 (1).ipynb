**Random Variable & Probability Distribution**

# Define a random variable for the event “Number of students passing the final exam” out of 3 randomly selected students.


```python

sample_space={'PPP','PPF','PFP','FPP','FFP','FPF','PFF','FFF'}
rv=set()
for i in sample_space:
  if 'PPP' in i:
    rv.add(3)
  elif 'PP' in i:
    rv.add(2)
  elif 'P' in i:
    rv.add(1)
  else:
    rv.add(0)
print(f'''according to our event which is number of students passed
the random variable is {rv} ''' )
```

# **Find the mean and variance of this random variable.**

```python
# as we already made probability distribution table in theory part
# from that table we will calculate mean and variance


mean=(0.446)+(2*0.239)+(3*0.0417)
print(f'the mean of random variable is {mean}')
variance=(((0-mean)**2)*0.2724+((1-mean)**2)*0.4469+((2-mean)**2)*0.2390+((3-mean)**2)*0.0417)
print(f'the variance of random variable is {variance:.2f}')
```

```python
# !pip install matplotlib-venn

from matplotlib_venn import venn2
import matplotlib.pyplot as plt

# Calculate the counts for each condition
study_hours_count = len(df[df['study_hours'] > 10])
attendance_count = len(df[df['attendance'] > 80])
both_count = len(df[(df['study_hours'] > 10) & (df['attendance'] > 80)])

# Calculate the unique counts for the Venn diagram
only_study_hours = study_hours_count - both_count
only_attendance = attendance_count - both_count

# Create the Venn diagram
plt.figure(figsize=(8,8))
venn2(subsets=(only_study_hours, only_attendance, both_count),set_labels=('Study Hours > 10', 'Attendance > 80'))
plt.title("Venn Diagram: Students with >10 Study Hours and >80 Attendance")
plt.show()
```
